### Вычисление специфичных для варианта переменных.

```python
import numpy as np
from dataclasses import dataclass
from pprint import pprint


N_VAR = 6

@dataclass(frozen=True)
class Part1:
    b0: np.float64
    b1: np.float64
    b2: np.float64
    a0: int
    a1: np.float64
    a2: np.float64
    N1: int
    N2: int
    fd: np.float64

@dataclass(frozen=True)
class Part2:
    N:  int
    fd: np.float64
    T:  np.float64
    A1: np.float64
    A2: np.float64
    f1: np.float64
    f2: np.float64

# Part 1
b0_1 = np.float64(0.5 + 0.02 * N_VAR)
b1_1 = np.float64(b0_1 * (-1)**(N_VAR + 1) * (0.9822 + 0.0178 * N_VAR))
b2_1 = np.float64(b0_1 * (0.8 + 0.2 * (N_VAR % 5)))
a0_1 = 1
a1_1 = np.float64(-1**N_VAR * (0.7778 + 0.025 * N_VAR))
a2_1 = np.float64(0.64 + 0.006 * N_VAR)
N1_1 = N_VAR % 10 + 20
N2_1 = N_VAR % 10 + 30
fd_1 = 1000 * N_VAR

part1 = Part1(
    b0=b0_1,
    b1=b1_1,
    b2=b2_1,
    a0=a0_1,
    a1=a1_1,
    a2=a2_1,
    N1=N1_1,
    N2=N2_1,
    fd=fd_1,
)

# Part 2
N_2  = 64
fd_2 = np.float64(2000 * (N_VAR % 5 + 1))
T_2  = np.float64(1 / fd_2)
A1_2 = np.float64(1 + 0.01 * N_VAR)
A2_2 = np.float64(2 * A1_2)
f1_2 = np.float64(fd_2 / 8)
f2_2 = np.float64(1.5 * f1_2)

part2 = Part2(
    N=N_2,
    fd=fd_2,
    T=T_2,
    A1=A1_2,
    A2=A2_2,
    f1=f1_2,
    f2=f2_2,
)

pprint(part1)
pprint(part2)
```
